<h1>Datasetin ymmärtäminen</h1>
<h2>Tunnistetut data tyypit</h2>

<h3>Nominal data</h3>
<p>
Kuvaus: Kuvastaa dataa, joiden arvojen väliä ei voi mitata tai niitä ei voi järjestää.<br>
Esimerkiksi:<br>
-värit. Emme voi todeta että minun lempivärini vihreä on parempi kuin sinun lempivärisi punainen.<br>
-pelaajanumerot. Emme voi todeta että pelaaja numero 8 on kaksi kertaa parempi pelaaja kuin 4 tai pelaaja 10 tulee listalla ennen pelaajaa 23.<br>
<br>
Nominal data on siis dataa, joka voi olla hyödyllistä/tärkeää, mutta emme voi lajitella tai laskea esimerkiksi keskiarvoa.
<br>
</p>

<h3>Ordinal data</h3>
<p>
Kuvaus: Kuvastaa dataa jossa on selkeä järjestys, mutta emme voi mitata välimatkoja niiden välillä<br>
Esimerkiksi:<br>
-Juoksukilpailu. Antti tulee sijalle 1, Juho sijalle 2 ja Henkka sijalle 3.<br>
Tarvitsisimme kuitenkin toisen datan kertomaan meille heidän maaliin saaapumis ajan, jolla voisimme mitata välimatkaa.<br>
-Elokuva arvostelu. Jos minä sanon että elokuva Goodfellas on 4/5 tähteä, mutta irishman vain 2/5.<br>
Se ei silti tee Goodfellas elokuvasta kaksi kertaa parempaa.<br>
Arvostelu tähdillä on hyvin suhteellista ja niiden väliä ei voi tarkkaan mitata<br>
<br>
</p>

<h3>Interval data</h3>
<p>
Kuvaus: Kuvastaa dataa, jossa meillä on järjestys ja suunta, mutta meiltä puuttu absoluuttinen nollapiste.<br>
Esimerkiksi:<br>
-Celcius aste. 0°C ei ole "ei ole lämpötilaa" vaan on oma lämpötilansa<br>
Emme voi myöskään sano että 50°C on puolet 100°C.<br>
Vaikka numeraalisesti 50 on puolet 100, mutta lämpötilassa se ei käy suoranaisesti järkeen<br>
Voimme tosin todeta että 100°C on lämpimämpi, kuin 50°C.<br>
<br>
</p>

<h3>Ratio data</h3>
<p>
Kuvaus: Kuvastaa dataa, jossa meillä on järjestys ja suunta, mutta meillä on myös absoluuttinen nollapiste.<br>
Esimerkiksi:<br>
-kelvin asteikko. Asteikon nollapiste on -273.15°C.<br>
Tämä mahdollistaa että voimme sanoa kelvin asteikon 100 olevan puolet 200:sta.<br>
-lapset. Jos sanoisin että minulla ei ole lapsia, joka on nollapiste.<br>
Tämä mahdollistaa että voimme sanoa että 4 lasta on tuplasti enemmän, kuin 2 lasta<br>
<br>
</p>

<h3>Binary data</h3>
<p>
Kuvaus: Kuvastaa dataa, jolla on kaksi mahdollista arvoa 0 & 1.<br>
Esimerkiksi:<br>
Titanicissa selvinneet on binäärinen arvo.<br>
<br>
</p>
