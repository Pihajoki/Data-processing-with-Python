```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(0, 5, 0.1)
y = np.sin(x)
fig = plt.figure()
ax1 = plt.subplot(2,2,1)
ax1.plot(x, y);
ax2 = plt.subplot(2,2,2)
ax2.plot(x, y);

ax3 = plt.subplot(2,2,3)
ax3.plot(x, y);
ax4 = plt.subplot(2,2,4)
ax4.plot(x, y);
```
