# Numpy examples with images

The purpose of these examples is just to show how numpy arrays come naturally in many practical situations, here with images.

### First test with images

```python
%matplotlib inline
#%matplotlib widget

import matplotlib.pyplot as plt
import numpy as np
```

```python
# For the sklearn you would need to install sci-kit-learn. Please see https://scikit-learn.org/stable/index.html
# we will also check this in class
from sklearn import datasets
digits = datasets.load_digits()
```

```python
digits.images[0]
```

```python
for i in range(3):
    plt.subplot(1, 3, i+1)
    plt.imshow(digits.images[i], cmap=plt.cm.gray_r)

plt.show()
```
