```python
df
```

```python
df[['c0', 'c1']]   # selecting columns "c0" and "c1". This we have seen already. Note the use of square brackets.
```

## Indexers loc and iloc

* loc operates with index labels
* iloc operates with index numeric numbers

#### Case Loc

```python
df.loc[['r0','r2'],['c0','c2']]                   # Rows first then columns
```

```python
df.loc[['r0','r2'],:] 
```

```python
df.loc[:,['c0','c2']] 
```

#### Case iloc

```python
df.iloc[[0,2],[0,2]]
# df.iloc[[0,2],['c0', 'c2']] => Does not work, requires numerical index also for columns!
```

```python
df.iloc[[0,2],:] 
```

```python
df.iloc[:,[0,2]]
```

```python
df.iloc[:,[0,2]]=1
```

```python
df
```

We can also create new columns into Pandas dataframe (table) easily. Here by calculating the sum of columns "c1" and "c2" and saving the result in new column called "c3"

```python
df['c3'] = df['c1'] + df['c2']
```

```python
df
```
